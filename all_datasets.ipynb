```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import *

def do_LinReg(X_cols, y_col):
  X_train, X_test, y_train, y_test = train_test_split(X_cols, y_col, test_size=0.2, random_state=42)

  model = LinearRegression()
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  print(f"MSE: {mean_squared_error(y_test, y_pred)}")
  print("R^2: ",r2_score(y_test, y_pred))

  coefficients = pd.DataFrame(model.coef_, X_cols.columns, columns=['Coefficient'])
  return coefficients

do_LinReg(df_sp500[['SP500']],df_sp500['PE10'])
```
